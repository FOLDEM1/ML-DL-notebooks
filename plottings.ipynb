```python
import numpy as np 
import matplotlib.pyplot as plt  
import matplotlib
import seaborn as sns 
from sklearn.datasets import load_iris 
import pandas as pd 
```

```python
X = np.linspace(0,10,100)
y = np.sin(X) + np.random.normal(0,.06,100)
```

```python
plt.scatter(X,y)
plt.show()
```

```python
plt.plot(X,y,marker='x',linestyle='--',linewidth=2,color='blue')
```

```python
fig = plt.figure(figsize=(10,5))
plt.scatter(X,y,s=20,marker='o',color='red',alpha=.5)
plt.plot(X,y,color='green',linestyle=":",linewidth=4)
plt.show()
```

```python

fig = plt.figure(figsize=(10,6),dpi=80)
plt.plot(X,y,marker='x', mec='r',color='g')
plt.xlim(0,10)
plt.ylim(-1.2,1.2)
plt.title('plot title',fontsize=20,loc='center')
plt.xlabel('X axis label',fontsize=15)
plt.ylabel('Y axis label',fontsize=15)
plt.grid(True,which='both',axis='both',color='gray',linestyle='--',linewidth=0.7)
plt.show()
```

```python



plt.subplot(1,2,1)
plt.plot(X,y,color='blue')
y=1/1+np.exp(-X)
plt.subplot(1,2,2)
plt.scatter(X,y,color='red')
```

```python
fig ,ax = plt.subplots(nrows=2,ncols=2)
fig.tight_layout()

y1 = np.cos(X)
y2 = np.sin(X)
y3 = np.tanh(X)
y4 = 1/ 1+np.exp(-X) 

ax[0,0].set_title('Cosine')
ax[0,1].set_title('Sine function')
ax[1,0].set_title('TanH')
ax[1,1].set_title("Sigmoid Function")

ax[0,0].plot(X,y1)
ax[0,1].plot(X,y2,color='red')
ax[1,0].plot(X,y3,color='green')
ax[1,1].plot(X,y4,color='orange')
fig.suptitle("Subploting suing fig + ax ")
plt.show()
```
