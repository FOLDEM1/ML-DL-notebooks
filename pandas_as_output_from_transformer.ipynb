```python
import numpy as np 
import pandas as pd 
```

```python
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
```

```python
data = pd.read_csv('penguins.csv')
```

```python
data
```

```python
data.dropna(inplace=True)
```

```python
data.drop(columns='id',inplace=True)
```

```python
from sklearn.pipeline import make_pipeline
```

```python

from sklearn import set_config
set_config(transform_output='pandas')
```

```python
y= data['sex']
X = data.drop(columns='sex')
```

```python
ct = make_column_transformer(
    (make_pipeline(StandardScaler()),X.select_dtypes(include='number').columns.values),
    (OneHotEncoder(sparse_output=False,drop='first'),['island','species']),
    remainder='passthrough',verbose_feature_names_out=True
)
```

```python
new_data = ct.fit_transform(data)
set_config(transform_output='default')
```

```python
ct
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier
```

```python
clf = make_pipeline(
    ct,HistGradientBoostingClassifier()
    
)
```

```python
clf
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,stratify=y,test_size=.25)
```

```python
clf.fit(X_train,y_train)
```

```python
clf.score(X_test,y_test)
```
